```python
import pandas as pd
```

```python
import numpy as np
```

```python
# Seed number
np.random.seed(42)
```

```python
# Create 150 instances of attributes for soccer players
completed_pass = np.random.randint(20, 100, size=150)
distance_traveled = np.round(np.random.uniform(8.0, 14.0, size=150), 1)
shots_on_goal = np.random.randint(0, 6, size=150)
disarms = np.random.randint(0, 15, size=150)
```

```python
# Positions divided equally of line players
positions_players = ['Striker'] * 50 + ['Midfielder'] * 50 + ['Defender'] * 50 
```

```python
np.random.shuffle(positions_players)
```

```python
dataset_players = {
    'completed_pass': completed_pass,
    'distance_traveled': distance_traveled,
    'shots_on_goal': shots_on_goal,
    'disarms': disarms,
    'positions_players': positions_players
}
```

```python
dataframe_soccer_players = pd.DataFrame(dataset_players)
```
